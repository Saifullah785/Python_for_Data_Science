```python
# importing necessary libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

<h2 style="color: #006064; margin-top: 0; font-weight: bold;">🎨 1. Colored Scatterplots</h2>
<p style="color: #263238; font-size: 15px; line-height: 1.8;">
How do we add a third dimension of information to a flat 2D scatter plot without making it complex? We use <b>Color Mapping</b>! Let's explore how we can visualize categories using the famous Iris dataset.
</p>
</div>

<div style="background: linear-gradient(135deg, #F9FBE7, #FFFDE7); border-left: 6px solid #C0CA33; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #558B2F; margin-top: 0;">📘 Definition: Group Representation via Color</h3>
<p style="color: #37474F; font-size: 14px; line-height: 1.7;">
In Matplotlib, we can assign a list of numerical labels to the <code>c</code> parameter and pair it with a color palette (<code>cmap</code>). This maps each class (e.g., flower species) to a distinct, easily recognizable color.
</p>
<p style="color: #558B2F; font-weight: bold; margin-bottom: 5px;">💡 Why It Matters</p>
<p style="color: #37474F; font-size: 14px; line-height: 1.7; margin-top: 0;">
Instead of presenting a monotone cloud of points, colored scatterplots instantly highlight underlying clusters and decision boundaries between groups.
</p>
<p style="color: #558B2F; font-weight: bold; margin-bottom: 5px;">🧠 Key Insight</p>
<p style="color: #37474F; font-size: 14px; line-height: 1.7; margin-top: 0;">
We convert textual classes like <code>Iris-setosa</code>, <code>Iris-versicolor</code>, and <code>Iris-virginica</code> into numerical codes (<code>0, 1, 2</code>) so Matplotlib's colormap engine can accurately match them to its spectrum.
</p>
</div>

```python
# loading the dataset

iris = pd.read_csv('iris.csv')
iris.sample(5)
```

```python
# assigning numerical values to species column

iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
# ploting the scatter plot

plt.figure(figsize=(10,7))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c= iris['Species'],cmap='jet',alpha=0.9)
plt.title('Scatter Plot')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

<h2 style="color: #E65100; margin-top: 0; font-weight: bold;">🏷️ 2. Annotations & Bubble Charts</h2>
<p style="color: #3E2723; font-size: 15px; line-height: 1.8;">
Sometimes, knowing <i>where</i> a point lies on a grid isn't enough—we need to know <b>who</b> or <b>what</b> that point represents. In this section, we'll annotate individual data points using a real-world dataset of cricket batters, mapping their total runs to the size of each bubble!
</p>
</div>

<div style="background: linear-gradient(135deg, #E8F5E9, #F1F8E9); border-left: 6px solid #4CAF50; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #1B5E20; margin-top: 0;">📘 Definition: Annotations & Sizing</h3>
<p style="color: #263238; font-size: 14px; line-height: 1.7;">
<b>Annotations</b> allow us to overlay textual descriptions directly onto plots using <code>plt.text()</code>. When paired with <b>Bubble Sizing</b> (using the <code>s</code> parameter), a single scatter plot can display four dimensions: X-position, Y-position, point size, and point label.
</p>
<p style="color: #1B5E20; font-weight: bold; margin-bottom: 5px;">🚀 Practical Example</p>
<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">
We plot cricket batters with <b>Average</b> on the X-axis and <b>Strike Rate</b> on the Y-axis. The size of their circle corresponds to their <b>total runs</b>, and each circle is clearly labeled with the player's name.
</p>
</div>

```python
# Loading another dataset, 'batter.csv', into a pandas DataFrame.
batter = pd.read_csv('batter.csv')
batter.head()
```

```python
# Taking a sample of 25 rows from the first 100 rows of the batter DataFrame.
# random_state ensures reproducibility of the sample.

sample_df = batter.head(100).sample(25,random_state=5)
```

```python
sample_df.sample(5)
```

```python
# Creating a figure with a specified size for the plot.

plt.figure(figsize=(18,10))

# Creating a scatter plot with 'avg' on the x-axis and 'strike_rate' on the y-axis.
# The size of the points is determined by the 'runs' column.

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

# Iterating through each row of the sample DataFrame to add annotations.

for i in range(sample_df.shape[0]):

  # Adding text annotations for each point.
  # The text is the value from the 'batter' column.
  # The position of the text is given by the 'avg' and 'strike_rate' values.

  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

<h2 style="color: #4A148C; margin-top: 0; font-weight: bold;">📏 3. Horizontal and Vertical Reference Lines</h2>
<p style="color: #263238; font-size: 15px; line-height: 1.8;">
When analyzing distributions, how do we distinguish outstanding performers from the average? We can draw explicit <b>benchmark lines</b>! Let's see how to add thresholds to our cricket analysis.
</p>
</div>

<div style="background: linear-gradient(135deg, #E0F2F1, #F4FBFB); border-left: 6px solid #009688; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #004D40; margin-top: 0;">📘 Definition: Reference Lines</h3>
<p style="color: #263238; font-size: 14px; line-height: 1.7;">
Using <code>plt.axhline()</code> (horizontal line) and <code>plt.axvline()</code> (vertical line), we can segment our plot into quadrant zones to represent benchmarks like a "high strike rate threshold" or a "target average".
</p>
<p style="color: #004D40; font-weight: bold; margin-bottom: 5px;">💡 Why It Matters</p>
<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">
It provides instant visual context. Anyone looking at the chart can immediately identify which players sit in the elite upper-right quadrant (High Average & High Strike Rate).
</p>
</div>

```python
# Defining lists for x and y coordinates.

x = [1,2,3,4]
y = [5,6,7,8]

# Creating a scatter plot of x and y.

plt.scatter(x,y)

# Adding text annotations to specific points on the plot.
# 'fontdict' is used to specify font properties like size and color.

plt.text(1,5,'point1',fontdict={'size':12, 'color': 'pink'})
plt.text(2,6,'point2',fontdict={'size':12, 'color': 'blue'})
plt.text(3,7,'point3',fontdict={'size':12, 'color': 'orange'})
plt.text(4,8,'point4',fontdict={'size':12, 'color': 'red'})
```

```python
plt.figure(figsize=(18,10))

# Creating a scatter plot with 'avg' on the x-axis and 'strike_rate' on the y-axis.
# The size of the points is determined by the 'runs' column.

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

# Adding horizontal lines to the plot at y=130 (red) and y=140 (green).

plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(30,color='red')

# Iterating through each row of the sample DataFrame to add annotations.

for i in range(sample_df.shape[0]):

   # Adding text annotations for each point, using values from the 'batter' column.

  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

<h2 style="color: #1A237E; margin-top: 0; font-weight: bold;">🖼️ 4. Subplots</h2>
<p style="color: #263238; font-size: 15px; line-height: 1.8;">
Sometimes, squeezing too much information into one graph causes clutter. The solution? <b>Subplots!</b> By splitting our canvas into a structured grid, we can tell multiple clean stories side-by-side.
</p>
</div>

<div style="background: linear-gradient(135deg, #FFF3E0, #FFFBF7); border-left: 6px solid #FF9800; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #E65100; margin-top: 0;">📘 Definition: Subplot Grids</h3>
<p style="color: #263238; font-size: 14px; line-height: 1.7;">
Using <code>plt.subplots(nrows, ncols)</code>, Matplotlib creates a structured matrix of axes (graphs) in a single figure window, which we can reference using indexes like <code>ax[0]</code> or <code>ax[0,1]</code>.
</p>
<p style="color: #E65100; font-weight: bold; margin-bottom: 5px;">🧠 Key Insight</p>
<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">
Sharing coordinates using <code>sharex=True</code> ensures that as we compare different metrics (such as Strike Rate vs. Runs) across the same independent variable, the scaling remains perfectly aligned for a fair comparison.
</p>
</div>

```python
# Displaying the first few rows of the 'batter' dataframe to see the data structure.
batter.head()
```

```python
# Creating a simple scatter plot of 'avg' vs 'strike_rate'.

plt.scatter(batter['avg'],batter['strike_rate'])
```

```python
# Creating a figure and a set of subplots.
# fig is the figure object, ax is an array of axes objects.
# In this case, we have one subplot.

fig, ax = plt.subplots(figsize=(15,10))
ax.scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax.set_title('Scatter Plot')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')
fig.show()
```

```python
batter.head()
```

```python
# Creating a figure and a grid of 2x1 subplots, sharing the x-axis.

fig, ax =plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(15,10))
ax[0].scatter(batter['avg'],batter['strike_rate'],color='red',marker='D')
ax[1].scatter(batter['avg'],batter['runs'],color='blue',marker='+')

ax[0].set_title('Average vs Strike Rate')
ax[0].set_xlabel('Average')
ax[0].set_ylabel('Strike Rate')

ax[1].set_xlabel('Average')
ax[1].set_ylabel('Runs')
ax[1].set_title('Average vs Runs')

fig.show()

```

```python
# Creating a figure and a grid of 2x2 subplots.

fig, ax =plt.subplots(nrows=2,ncols=2,figsize=(10,10))
ax[0,0].scatter(batter['avg'],batter['strike_rate'])
ax[0,1].scatter(batter['avg'],batter['runs'],color='blue',marker='+')
ax[1,0].hist(batter['avg'],color='green')
ax[1,1].hist(batter['runs'],color='orange')



fig.show()
```

```python
# Another way to create subplots using fig.add_subplot().
# Creating a figure.

fig = plt.figure(figsize=(10,10))
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'])
ax2 = fig.add_subplot(2,2,2)
ax2.scatter(batter['avg'],batter['runs'],color='blue',marker='+')
ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'],color='green')
ax4 = fig.add_subplot(2,2,4)
ax4.hist(batter['runs'],color='orange')



fig.show()
```

By declaring <code>projection='3d'</code> inside our subplots, we activate Matplotlib's 3D toolkit (<code>mplot3d</code>). This allows us to plot <code>x</code>, <code>y</code>, and <code>z</code> data using <code>scatter3D()</code> and <code>plot3D()</code>.

<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">

```python
batter
fig = plt.figure()

ax = plt.subplot(projection='3d')
ax.scatter3D(batter['avg'],batter['strike_rate'],batter['runs'])
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')
ax.set_zlabel('Runs')
plt.show()
```

<h2 style="color: #004D40; margin-top: 0; font-weight: bold;">🏔️ 6. 3D Surface Plots</h2>
<p style="color: #263238; font-size: 15px; line-height: 1.8;">
To visualize mathematical relationships, optimization landscapes, or physical terrains, we can construct <b>continuous 3D surfaces</b>. Let's explore how coordinate grids construct these topography-like visualizations.
</p>
</div>

<div style="background: linear-gradient(135deg, #E8F5E9, #F3FAF4); border-left: 6px solid #4CAF50; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #1B5E20; margin-top: 0;">📘 Definition: Meshgrids & continuous surfaces</h3>
<p style="color: #263238; font-size: 14px; line-height: 1.7;">
Using <code>np.meshgrid()</code>, we convert 1D coordinate arrays into 2D matrices representing every combination of X and Y. We then compute the functional values of Z across this grid and plot them using <code>ax.plot_surface()</code>.
</p>
<p style="color: #1B5E20; font-weight: bold; margin-bottom: 5px;">🧠 Key Insight</p>
<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">
Adding a colormap (like <code>cmap='viridis'</code>) is crucial for surface plots. It provides a visual indicator of height, helping the viewer interpret peaks and valleys instantly.
</p>
</div>

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
x,y = np.meshgrid(x,y)
z = x**2 + y**2
```

```python
z.shape
```

```python
fig = plt.figure(figsize=(10,10))
ax = plt.subplot(projection='3d')
p=ax.plot_surface(x,y,z,cmap='viridis')
fig.colorbar(p)

ax.set_title('3D Surface Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
```

```python
z = np.sin(x) + np.cos(y)
fig = plt.figure(figsize=(10,10))
ax = plt.subplot(projection='3d')
p=ax.plot_surface(x,y,z,cmap='viridis')
fig.colorbar(p)

ax.set_title('3D Surface Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
```

<h2 style="color: #F57F17; margin-top: 0; font-weight: bold;">🌐 7. Contour Plots</h2>
<p style="color: #263238; font-size: 15px; line-height: 1.8;">
How do we represent a 3D landscape on a completely flat 2D sheet of paper? We use **Contour Plots**—the exact same technique used by topographers in geographic maps!
</p>
</div>

<div style="background: linear-gradient(135deg, #E1F5FE, #F2FAFD); border-left: 6px solid #0288D1; border-radius: 12px; padding: 18px; margin: 15px 0; box-shadow: 0 4px 10px rgba(0,0,0,0.05);">
<h3 style="color: #01579B; margin-top: 0;">📘 Definition: Isolines & Filled Contours</h3>
<p style="color: #263238; font-size: 14px; line-height: 1.7;">
<b>Contour lines</b> (<code>ax.contour()</code>) connect points of equal height/value (Z). <b>Filled contours</b> (<code>ax.contourf()</code>) color the spaces between these lines to create smooth gradients of value change.
</p>
<p style="color: #01579B; font-weight: bold; margin-bottom: 5px;">💡 Why It Matters</p>
<p style="color: #263238; font-size: 14px; line-height: 1.7; margin-top: 0;">
In machine learning, contour plots are used to visualize loss landscapes and decision boundaries, allowing us to see how optimization algorithms navigate down towards global minima.
</p>
</div>

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
x,y = np.meshgrid(x,y)
z = x**2 + y**2
```

```python


fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(x,y,z,cmap='viridis')

fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(x,y,z,cmap='viridis')

fig.colorbar(p)
```

```python
z = np.sin(x) + np.cos(y)
fig = plt.figure(figsize=(10,10))
ax = plt.subplot()
p=ax.contourf(x,y,z,cmap='viridis')
fig.colorbar(p)

ax.set_title('3D Surface Plot')
```

```python
# pie -> single and multiple
df = pd.DataFrame(
    {
        'batsman':['Dhawan','Rohit','Kohli','SKY','Pandya','Pant'],
        'match1':[120,90,35,45,12,10],
        'match2':[0,1,123,130,34,45],
        'match3':[50,24,145,45,10,90]
    }
)

df.head()
```

```python
df['match1'].plot(kind='pie',labels=df['batsman'],autopct='%1.1f%%',explode=[0,0.1,0,0,0,0])
```

```python
# # multiple pie charts
# df.iloc[:,1:].plot(kind='pie',subplots=True,figsize=(10,10))

# multiple pie charts

df[['match1','match2','match3']].plot(kind='pie',subplots=True,figsize=(15,8))
```

```python
# multiple separate graphs together using stock
stocks.plot(kind='line',figsize=(10,10))
stocks.plot(kind='line',subplots=True,figsize=(10,10))
```

```python
# on multiindex dataframes
# using tips

tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind='bar')
```

```python
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind='pie',subplots=True,figsize=(20,10))
```
