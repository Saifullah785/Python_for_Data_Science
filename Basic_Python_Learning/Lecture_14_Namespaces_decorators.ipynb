# Namespaces

A namespace is a space that holds names(identifiers). Programmatically speaking, namespaces are dictionay of identifiers(keys) and their objects(values)

# There are 4 types of namespaces:

- Builtin Namespace

- Global Namespace

- Enclosing Namespace

- local Namespaces


# Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in-scope.

if the interpreter doesn't find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global
# global var

a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> same name

a = 2

def temp():
  # local var
  a = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> local does not have but global has

a = 2

def temp():
  # local var
  print(a)

temp()
print(a)
```

```python
# local and global -> editing global
a = 2

def temp():
  # local var
  a += 1
  print(a)

temp()
print(a)
```

```python
a = 2

def temp():
  # local var
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# local and global -> global created inside local
def temp():
  # local var
  global a
  a = 1
  print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z):
  # local var
  print(z)

a = 5
temp(5)
print(a)
print(z)
```

# Decorators

A decorator in Python is a function that receives another function as input an adds some functionality(decoration) to and it and return it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

- Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc

- User defined decorators that we programmers can create according to our needs

```python
# Python are 1st class function
def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
# simple example
def my_decorator(func):
  def wrapper():
    print('*****************************')
    func()
    print('*****************************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello saif')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```
