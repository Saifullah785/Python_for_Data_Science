<p><b>Example:</b> <code>7</code> is an <code>int</code>, <code>7.5</code> is a <code>float</code>, and <code>'hello'</code> is a <code>str</code>. You can check a variable's type using <code>type()</code>.</p>

<p><b>Key Idea:</b> Python's flexible data types allow it to represent diverse kinds of information, crucial for different programming tasks.</p>

</div>

# 3.Variable

<div style="background-color:#F5F3FF; border-left:5px solid #8B5CF6; border-radius:10px; padding:18px; margin:15px 0;">

<h3>🏷️ Variables in Python</h3>

<p><b>Definition:</b> A variable is a name that refers to a value or object stored in memory. It acts as a label, allowing you to store, access, and manipulate data throughout your program.</p>

<p><b>Example:</b> <code>name = 'saif'</code> assigns the string 'saif' to the variable <code>name</code>. Later, <code>age = 30</code> assigns the integer 30 to <code>age</code>. Python supports <b>dynamic typing</b> (types are determined at runtime) and <b>dynamic binding</b> (a variable's type can change).</p>

<p><b>Key Idea:</b> Variables make your code reusable and readable by giving meaningful names to values.</p>

</div>

```python
# This code assigns the string 'saif' to the variable 'name' and then prints the value of the variable.
name = 'saif'

print(name)
```

```python
# This code assigns integer values to variables 'a' and 'b', then prints the sum of 'a' and 'b'.
a = 4
b = 5

print(a + b)
```

```python
# This is an example of dynamic typing in Python, where the data type of a variable is determined at runtime.
a = 5

# This is commented out as it represents static typing, which is not how variable declaration works in Python.
# static typing
# int a = 5
```

```python
# This is an example of dynamic binding in Python, where the type of the object a variable refers to can change during runtime.
# Initially, 'a' is an integer.
a = 4
print(a)

# Then, 'a' is reassigned to a string.
a = 'ali'
print(a)
```

```python
# This code assigns integer values to variables a, b, and c individually and then prints their values.
a = 1
b = 2
c = 3
print(a,b,c)
```

```python
# This code demonstrates assigning multiple values to multiple variables in a single line (tuple unpacking).
a,b,c =1,2,3
print(a,b,c)
```

```python
# This code demonstrates assigning the same value to multiple variables in a single line.
a= b= c = 5
print(a,b,c)
```

<h3>🔑 Keywords & Identifiers</h3>

<p><b>Definition:</b> <b>Keywords</b> are special reserved words in Python that have predefined meanings and cannot be used as variable names. <b>Identifiers</b> are names given to variables, functions, classes, etc., created by the programmer.</p>

<p><b>Key Idea:</b> Understanding keywords and identifier rules is fundamental for writing syntactically correct and meaningful Python code.</p>

</div>

# 7.Type Conversion

<div style="background-color:#F8FAFC; border-left:5px solid #4F46E5; border-radius:10px; padding:18px; margin:15px 0;">

<h3>🔄 Type Conversion</h3>

<p><b>Definition:</b> Type conversion (or type casting) is the process of changing a value from one data type to another. This can be <b>implicit</b> (done automatically by Python) or <b>explicit</b> (done manually by the programmer).</p>

<p><b>Example:</b> <code>4 + 5.6</code> results in <code>9.6</code> (implicit conversion of <code>4</code> to <code>float</code>). <code>int('5')</code> explicitly converts the string '5' to the integer 5, while <code>float(10)</code> converts 10 to 10.0.</p>

<p><b>Key Idea:</b> Type conversion is essential when operations require specific data types or when handling user input that needs to be treated as numbers.</p>

</div>

```python
# implicit Vs Explicit
print(5 + 5.6)

print(type(5),type(5.6))

a = float('5.0')
print(4 + a)
```

# 8.Literals

<div style="background-color:#F0F9FF; border-left:5px solid #0EA5E9; border-radius:10px; padding:18px; margin:15px 0;">

<h3>✨ Literals in Python</h3>

<p><b>Definition:</b> Literals are raw data values or fixed values that are directly embedded in your code. They are used to represent specific values like numbers, strings, or boolean states.</p>

<p><b>Example:</b> <code>100</code> is an integer literal, <code>3.14</code> is a float literal, <code>'hello'</code> is a string literal, <code>True</code> is a boolean literal. Python also supports binary (<code>0b...</code>), octal (<code>0o...</code>), and hexadecimal (<code>0x...</code>) integer literals, as well as complex number literals (<code>3+4j</code>).</p>

<p><b>Key Idea:</b> Literals are the basic building blocks for representing concrete data within your Python programs.</p>

</div>

```python
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5
float_2 = 1.5e2 #= 1.5 * 10^2
float_3 = 1.5e-3 #= 1.5 * 10^-3

#Complex Literal
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
```
