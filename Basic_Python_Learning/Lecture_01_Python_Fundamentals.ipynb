# 3. Variable

```python
# This code assigns the string 'saif' to the variable 'name' and then prints the value of the variable.
name = 'saif'

print(name)
```

```python
# This code assigns integer values to variables 'a' and 'b', then prints the sum of 'a' and 'b'.
a = 4
b = 5

print(a + b)
```

```python
# This is an example of dynamic typing in Python, where the data type of a variable is determined at runtime.
a = 5

# This is commented out as it represents static typing, which is not how variable declaration works in Python.
# static typing
# int a = 5
```

```python
# This is an example of dynamic binding in Python, where the type of the object a variable refers to can change during runtime.
# Initially, 'a' is an integer.
a = 4
print(a)

# Then, 'a' is reassigned to a string.
a = 'ali'
print(a)
```

```python
# This code assigns integer values to variables a, b, and c individually and then prints their values.
a = 1
b = 2
c = 3
print(a,b,c)
```

```python
# This code demonstrates assigning multiple values to multiple variables in a single line (tuple unpacking).
a,b,c =1,2,3
print(a,b,c)
```

```python
# This code demonstrates assigning the same value to multiple variables in a single line.
a= b= c = 5
print(a,b,c)
```
