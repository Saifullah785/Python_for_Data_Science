## Accessing Substrings from a String

```python
# Positive Indexing
s = 'hello world'
print(s[0])
print(s[1])
```

```python
# Negative indexing

s = 'hello world'
print(s[-1])
print(s[-2],s[-3])
```

```python
# silicing

s = 'hello world'
print(s[0:5])
print(s[0:6:2])
print(s[6:0:-2])
```

```python
print(s[::-1])
```

```python
s = 'hello world'
print(s[-5:])
print(s[-1:-6:-1])
```
