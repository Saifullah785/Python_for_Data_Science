# Namespaces

A namespace is a space that holds names(identifiers). Programmatically speaking, namespaces are dictionay of identifiers(keys) and their objects(values)

# There are 4 types of namespaces:

- Builtin Namespace

- Global Namespace

- Enclosing Namespace

- local Namespaces

  <h3 style="color:#1D4ED8; margin-top:0;">📘 Understanding Namespaces</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> A <strong>Namespace</strong> is essentially a system Python uses to ensure that all the names (identifiers) in a program are unique and can be resolved without conflict. Under the hood, Python implements namespaces as standard dictionary mappings, where the dictionary keys represent variable/function names, and the values point to the actual memory objects.</p>

# Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in-scope.

if the interpreter doesn't find the name in any of these locations, then Python raises a NameError exception.

<div style="background-color:#F5F3FF; border:1px solid #DDD6FE; border-left:6px solid #7C3AED; border-radius:12px; padding:18px 20px; margin:18px 0; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;">
  <h3 style="color:#6D28D9; margin-top:0;">🔮 The LEGB Scope Resolution Rule</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> <strong>Scope</strong> defines the textual area of your code where a specific namespace is directly accessible. When you reference an identifier, Python searches for it following a strict hierarchical sequence known as <strong>LEGB</strong>:</p>
  
  <ol style="color:#334155; line-height:1.6; padding-left:20px;">
    <li><strong>L (Local):</strong> Names assigned inside a function (or lambda) and not declared global.</li>
    <li><strong>E (Enclosing):</strong> Names in the local scope of any and all outer enclosing functions (from the inner-most outward).</li>
    <li><strong>G (Global):</strong> Names assigned at the top-level of a module file, or declared global in a def.</li>
    <li><strong>B (Built-in):</strong> Pre-loaded names in Python (like <code>print</code>, <code>max</code>, <code>ValueError</code>).</li>
  </ol>
  
  <p style="color:#1E293B; margin-bottom:0;"><strong>⚠️ If all lookup steps fail:</strong> Python raises a <code>NameError</code> exception.</p>
</div>

```python
# local and global
# global var

a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> same name

a = 2

def temp():
  # local var
  a = 3
  print(b)

temp()
print(a)
```

  <h3 style="color:#B45309; margin-top:0;">⚠️ The UnboundLocalError Trap</h3>
  <p style="color:#1E293B;"><strong>The Mechanism:</strong> When Python parses a function body and detects a variable assignment (like <code>a += 1</code> or <code>a = 3</code>), it automatically tags that variable as <strong>local</strong> to that function. If you try to read from it before assigning a value inside that function, Python throws an <code>UnboundLocalError</code>, even if a global variable with the same name exists.</p>
</div>

```python
# local and global -> editing global
a = 2

def temp():
  # local var
  a += 1
  print(a)

temp()
print(a)
```

  <h3 style="color:#047857; margin-top:0;">🛠️ Overriding Local Isolation: The global Keyword</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> The <code>global</code> keyword explicitly instructs Python to map an identifier directly to the global module-level namespace, allowing inside-function modifications to persist globally.</p>
</div>

```python
# This cell demonstrates using the 'global' keyword to modify a global variable within a function.
a = 2

def temp():
  # local var
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# local and global -> global created inside local
def temp():
  # local var
  global a
  a = 1
  print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z):
  # local var
  print(z)

a = 5
temp(5)
print(a)
print(z)
```

# Decorators

A decorator in Python is a function that receives another function as input an adds some functionality(decoration) to and it and return it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

- Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc

- User defined decorators that we programmers can create according to our needs

<div style="background-color:#EFF6FF; border:1px solid #BFDBFE; border-left:6px solid #2563EB; border-radius:12px; padding:18px 20px; margin:18px 0; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;">
  <h3 style="color:#1D4ED8; margin-top:0;">🥇 First-Class Functions: The Gateway to Decorators</h3>
  <p style="color:#1E293B;">In Python, functions are <strong>First-Class Citizens</strong>. This means they can be treated like any other data type: they can be passed as arguments to other functions, returned from functions, and assigned to variables.</p>
</div>

```python
# Python are 1st class function
def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

<div style="background-color:#F5F3FF; border:1px solid #DDD6FE; border-left:6px solid #7C3AED; border-radius:12px; padding:18px 20px; margin:18px 0; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;">
  <h3 style="color:#6D28D9; margin-top:0;">🎬 Dynamic Modification with Decorators</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> A <strong>Decorator</strong> wraps another function to dynamically extend or modify its execution path without permanently rewriting its source code.</p>
</div>

```python
# simple example
def my_decorator(func):
  def wrapper():
    print('*****************************')
    func()
    print('*****************************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello saif')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

  <h3 style="color:#047857; margin-top:0;">✨ Elegant Syntactic Sugar: The @ Symbol</h3>
  <p style="color:#1E293B;">Instead of manually re-binding your functions like <code>hello = my_decorator(hello)</code>, Python offers a cleaner syntax using the <code>@decorator_name</code> operator placed directly above the definition.</p>
</div>

```python
# better syntax
# simple example

def my_decorator(func):
  def wrapper():
    print('*****************************')
    func()
    print('*****************************')
  return wrapper

@my_decorator
def hello():
  print('hello')
```

```python
# anything meaningful?
import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by', func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello world')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```

  <h3 style="color:#B45309; margin-top:0;">🛡️ Advanced Decorators with Configurations</h3>
  <p style="color:#1E293B;">If a decorator needs to accept arguments (like data types or limits), we wrap the wrapper inside another top-level function that captures those variables, returning the configured decorator structure dynamically.</p>

```python
# This cell defines a 'square' function and applies the 'sanity_check' decorator to validate its input type.
@sanity_check(int)
def square(num):
  print(num**2)
```

```python
# This cell defines the 'sanity_check' decorator to validate function argument types.
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```
