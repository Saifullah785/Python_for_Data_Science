#**Operators in Pyhton**

- Arithmetic Operators

- Relational Operators

- Logical Operators

- Bitwise Operators

- Assignment Operators

- Membership Operators

```python
# Demonstrates arithmetic operators in Python
# Arithmetic Operators

print(5+6) # Prints the sum of 5 and 6

print(5-6) # Prints the difference of 5 and 6

print(5*6) # Prints the product of 5 and 6

print(5/2) # Prints the division of 5 by 2 (float division)

print(5//2) # Prints the division of 5 by 2 (integer division)

print(5%2) # Prints the remainder of 5 divided by 2

print(5**3) # Prints 5 raised to the power of 3
```

```python
# Demonstrates relational operators in Python
# Relational Opertors

print(4>5) # Checks if 4 is greater than 5

print(4<5) # Checks if 4 is less than 5

print(4>=4) # Checks if 4 is greater than or equal to 4

print(4<=4) # Checks if 4 is less than or equal to 4

print(4==4) # Checks if 4 is equal to 4

print(4!=4) # Checks if 4 is not equal to 4
```

```python
# Demonstrates logical operators in Python
# Logical Operators

print(1 and 0) # Performs a logical AND operation

print(1 or 0) # Performs a logical OR operation

print(not 1) # Performs a logical NOT operation
```

```python
# Demonstrates bitwise operators in Python
# Bitwise Operators

print(2 & 3) # Performs a bitwise AND operation

print(2 | 3) # Performs a bitwise OR operation

print(2 ^ 3) # Performs a bitwise XOR operation

print(~3) # Performs a bitwise NOT operation

print(4 >> 2) # Performs a bitwise right shift operation

print(5 << 2) # Performs a bitwise left shift operation
```

```python
# Demonstrates assignment operators in Python
# Assignment Operators

a = 2 # Assigns the value 2 to variable a

print(a) # Prints the value of a

# a = a +2
# print(a)
a += 3 # Adds 3 to a and assigns the result back to a

print(a) # Prints the value of a

a -= 4 # Subtracts 4 from a and assigns the result back to a

print(a) # Prints the value of a

a *= 5 # Multiplies a by 5 and assigns the result back to a

print(a) # Prints the value of a

a /= 6 # Divides a by 6 and assigns the result back to a

print(a) # Prints the value of a

a %= 7 # Calculates the remainder of a divided by 7 and assigns the result back to a

print(a) # Prints the value of a
```

```python
# Demonstrates membership operators in Python
# Membership Operators

print('D' in 'Delhi') # Checks if 'D' is present in the string 'Delhi'

print('M' in 'Delhi') # Checks if 'M' is present in the string 'Delhi'

print('D' not in 'Delhi') # Checks if 'D' is not present in the string 'Delhi'

print(1 in [2,3,4,5,6]) # Checks if 1 is present in the list [2,3,4,5,6]
```

# if-else in Python

```python
# Prompts the user to enter their email
email = input('Enter email: ')

# Prompts the user to enter their password
password = input('Enter password: ')

# Checks if the email and password match
if email == 'saif@gmail.com' and password == '1234':

# Prints a welcome message if the credentials are correct
  print('welcome Saifullah')

# Checks if the email is correct but the password is not
elif email == 'saif@gmail.com' and password != '1234':


# Informs the user that the password is incorrect
  print('Incorrect password Please try again')

 # Prompts the user to enter the password again
  password = input('Enter password again: ')

# Checks if the re-entered password is correct
  if password == '1234':

 # Prints a message if the password is correct
    print('Finally correct')

# If the re-entered password is still incorrect
  else:


# Informs the user that the password is still incorrect
    print('Still incorrect')


# If the email is incorrect
else:

# Informs the user that the email is incorrect and to try again
  print('Incorrect Please try again')
```

```python
fnum = int(input('Enter first number'))

snum = int(input('Enter second number'))

op = input('Enter operator')

if op == '+':

  print(fnum + snum)

elif op == '-':

  print(fnum - snum)

elif op == '*':

  print(fnum * snum)

else:
  print(fnum / snum)
```
