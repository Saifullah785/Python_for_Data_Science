#**Operators in Pyhton**

- Arithmetic Operators

- Relational Operators

- Logical Operators

- Bitwise Operators

- Assignment Operators

- Membership Operators

```python
# Demonstrates arithmetic operators in Python
# Arithmetic Operators

print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)

print(5%2)

print(5**3)
```

```python
# Demonstrates relational operators in Python
# Relational Opertors

print(4>5)

print(4<5)

print(4>=4)

print(4<=4)

print(4==4)

print(4!=4)
```

```python
# Demonstrates logical operators in Python
# Logical Operators

print(1 and 0)

print(1 or 0)

print(not 1)
```

```python
# Demonstrates bitwise operators in Python
# Bitwise Operators

print(2 & 3)

print(2 | 3)

print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
```

```python
# Demonstrates assignment operators in Python
# Assignment Operators

a = 2

print(a)

# a = a +2
# print(a)
a += 3

print(a)

a -= 4

print(a)

a *= 5

print(a)

a /= 6

print(a)

a %= 7

print(a)
```

```python
# Demonstrates membership operators in Python
# Membership Operators

print('D' in 'Delhi')

print('M' in 'Delhi')

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
```
