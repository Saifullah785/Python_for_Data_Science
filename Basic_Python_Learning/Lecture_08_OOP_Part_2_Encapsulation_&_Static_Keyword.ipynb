- A user can find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given  2D point and a given line

<div style="background-color:#F0F9FF;border-left:5px solid #0284C7;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#075985;">└─│ Practical Program: 2D Geometry Engine</h3>
<p><strong>└─│ Goal:</strong> Design standard mathematical entities (Point and Line) to calculate planar coordinates, Euclidean distances, and test spatial line intersections using vector projections.</p>
<p><strong>└─│ Mathematical Approach:</strong>
<ul>
<li><strong>Euclidean Distance:</strong> \(d = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}\)</li>
<li><strong>Distance from Origin:</strong> \(d = \sqrt{x^2 + y^2}\)</li>
<li><strong>Point-to-Line Distance:</strong> \(d = \frac{|Ax_0 + By_0 + C|}{\sqrt{A^2 + B^2}}\) for line \(Ax + By + C = 0\)</li>
</ul>
</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> Creating abstract mathematical data types using classes encapsulates coordinates alongside the mathematical methods acting on them.</p>
</div>

```python
# Defines a Point class to represent 2D coordinates
class Point:

  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cod,self.y_cod)


  def euclidean_distance(self,other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return (self.x_cod**2 + self.y_cod**2)**0.5

    # return self.education_distance(point(0,0))
```

```python
# Defines a Line class to represent a line in 2D space
class Line:
  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A,self.B,self.C)

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return "lies on the line"

    else:
      return "does not lie on the line"
  def shortest_distance(line,point):
    return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5
```

```python
# Creates Line and Point objects and tests the Line class methods
l1 = Line(1,1,-2)
p1 = Point(1,10)
print(l1)
print(p1)
print(l1.shortest_distance(p1))
print(l1.point_on_line(p1))
```

# How objects access attributes

<div style="background-color:#F5F3FF;border-left:5px solid #7C3AED;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#5B21B6;">└─│ Concept: Object Attributes and Methods</h3>
<p><strong>Definition:</strong> Attributes represent the state/data stored inside an object, accessed via <code>object.attribute</code>. Methods represent behaviors or functions stored inside a class, accessed via <code>object.method()</code>.</p>
<p><strong>└─│ Example:</strong> In the <code>Person</code> class, <code>name</code> is an attribute initialized within the constructor, whereas <code>greet()</code> is a method that inspects that state to print custom greetings.</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> Attributes define what an object <i>is</i>, while methods define what an object <i>does</i>.</p>
</div>

```python
# Defines a Person class to demonstrate object attributes and methods
class Person:
  def __init__(self,name_input,country_input):
    self.name = name_input
    self.country = country_input

  def greet(self):
    if self.country == 'Pakistan':
      print('Assalam_o_aliakum',self.name)
    else:
      print('hello',self.name)
```

```python
# Creates a Person object and accesses its attributes
# how to access attributes
p = Person('Ali','Pakistan')
print(p.name)
```

```python
# Calls the greet method of the Person object
p.greet()
```

```python
# Attempts to access a non-existent attribute of the Person object
p.gender
```

# Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variables
- Assigning a new reference variable to an existing object does not create a new object

<div style="background-color:#F0F9FF;border-left:5px solid #0284C7;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#075985;">└─│ Concept: Reference Variables & Memory Pointers</h3>
<p><strong>Definition:</strong> Variables in Python do not hold the actual object values. Instead, they store a memory address referencing the place where the actual object resides. The <code>id()</code> function displays this unique address during the lifetime of the object.</p>
<p><strong>└─│ Mechanics:</strong> Assigning <code>q = p</code> makes both variables point to the same object. Changes made via reference <code>q</code> are immediately visible when reading through reference <code>p</code>.</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> Assigning a reference to an existing object shares access; it does not duplicate the underlying object.</p>
</div>

```python
# Demonstrates object referencing
# object without a reference
class Person:
  def __init__(self):
    self.name = 'Ali'
    self.gender = 'male'

p = Person()
q = p
```

```python
# Prints the memory addresses of the two reference variables
# Multiple ref
print(id(p))
print(id(q))
```

```python
# Demonstrates changing attribute values using different reference variables
# change attribute value with the help of 2nd object
```

```python
# Prints the name attribute using both reference variables before and after modification
print(p.name)
print(q.name)
q.name = 'ankit'
print(q.name)
print(p.name)
```

# Pass By Reference

<div style="background-color:#FFFBEB;border-left:5px solid #D97706;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#92400E;">└─│ Concept: Pass By Reference (Call-by-Sharing)</h3>
<p><strong>Definition:</strong> Python passes object arguments to functions using <i>pass-by-sharing</i>. The function parameters act as local reference variables pointing to the same external object in memory.</p>
<p><strong>└─│ Behavior:</strong>
<ul>
<li>Modifying object properties inside the function affects the original caller's object.</li>
<li>Reassigning the function's argument variable to a new object does not rebind the caller's external variable.</li>
</ul>
</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> User-defined objects behave as mutable structures; sharing them in functions exposes them to in-place updates.</p>
</div>

```python
# Demonstrates pass by reference with a function that modifies an object and returns a new one
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# outside the class -> function
def greet(person):
  print('Hi my name is',person.name,'and I am a',person.gender)
  p1 = Person('ankit','male')
  return p1

p = Person('Ali','male')
x = greet(p)
print(x.name)
print(x.gender)
```

```python
# Demonstrates pass by reference with a function that modifies an object in place
class Person:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

def greet(person):
  print(id(person))
  person.name = 'ankit'
  print(person.name)

p = Person('Ali','male')
print(id(p))
greet(p)
print(p.name)
```

#Encapsulation

<div style="background-color:#F5F3FF;border-left:5px solid #7C3AED;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#5B21B6;">└─│ Concept: Encapsulation & Data Hiding</h3>
<p><strong>Definition:</strong> Encapsulation is the practice of restricting direct access to object attributes. In Python, prefixing attributes with double underscores (e.g., <code>__balance</code>) triggers name mangling, transforming the attribute identifier into <code>_ClassName__attributeName</code> to prevent accidental modifications.</p>
<p><strong>└─│ Interface Pattern (Getters &amp; Setters):</strong> Controlled read/write interactions with protected attributes should be routed through dedicated getter and setter methods to enforce integrity rules.</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> Name mangling is a defensive layout mechanism against direct attribute modification, not an iron-clad security container.</p>
</div>

```python
# Demonstrates encapsulation by accessing public attributes
class Person:
  def __init__(self,name_input,country_input):
    self.name = name_input
    self.country = country_input

p1 = Person('Ali','Pakistan')
print(p1.name)
p2 = Person('Ankit','India')
print(p2.name)
```

# Collection of Objects

<div style="background-color:#F0FDFA;border-left:5px solid #0D9488;border-radius:10px;padding:18px;margin:14px 0;">
<h3 style="margin-top:0;color:#0F766E;">└─│ Concept: Collections of Objects</h3>
<p><strong>Definition:</strong> Python structures such as lists, tuples, and dictionaries can seamlessly store references to custom class instances, allowing you to iterate over object groups and call methods systematically.</p>
<p><strong>└─│ Practical Example:</strong> Constructing <code>L = [p1, p2, p3]</code> lets you loop through diverse <code>Person</code> records and programmatically display names and genders.</p>
<p style="margin-bottom:0;"><strong>└─│ Key Idea:</strong> Storing custom objects inside standard lists yields powerful database-style loops and batch manipulations.</p>
</div>

```python
# Creates a list of Person objects and iterates through it
class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

p1 = Person('Ali','male')
p2 = Person('Ankit','male')
p3 = Person('Neha','female')

L = [p1,p2,p3]

for i in L:
  print(i.name,i.gender)
```

# Static Variables(VS Instance Variables)

<div style="background-color:#FFFBEB;border-left:5px solid #D97706;border-radius:10px;padding:18px;margin:14px 0;">

```python
# Defines an ATM class with a static variable and method
class Atm:

  __counter = 1

  # constructor(special function)->superpower ->
  def __init__(self):
    print(id(self))
    self.pin = ''
    self.__balance = 0
    self.cid = Atm.__counter
    Atm.__counter = Atm.__counter + 1
    #self.menu()

  # utility functions
  @staticmethod
  def get_counter():
    return Atm.__counter


  def get_balance(self):
    return self.__balance

  def set_balance(self,new_value):
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print('beta bahot maarenge')

  def __menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input('enter your pin')
    self.pin = user_pin

    user_balance = int(input('enter balance'))
    self.__balance = user_balance

    print('pin created successfully')

  def change_pin(self):
    old_pin = input('enter old pin')

    if old_pin == self.pin:
      # let him change the pin
      new_pin = input('enter new pin')
      self.pin = new_pin
      print('pin change successful')
    else:
      print('nai karne de sakta re baba')

  def check_balance(self):
    user_pin = input('enter your pin')
    if user_pin == self.pin:
      print('your balance is ',self.__balance)
    else:
      print('chal nikal yahan se')

  def withdraw(self):
    user_pin = input('enter the pin')
    if user_pin == self.pin:
      # allow to withdraw
      amount = int(input('enter the amount'))
      if amount <= self.__balance:
        self.__balance = self.__balance - amount
        print('withdrawl successful.balance is',self.__balance)
      else:
        print('abe garib')
    else:
      print('sale chor')
```

#Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
-The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

```python
# Demonstrates static variables and methods with a Lion class
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```
