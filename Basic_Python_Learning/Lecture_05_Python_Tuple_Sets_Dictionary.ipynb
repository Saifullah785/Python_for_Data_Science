# Tuples

A tuple in Python is similar to a list.The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list.A tuple can not be changed in any way once it is created.

Characterstics

- Ordered
- Unchangeble
- Allows duplicate

## Plan of attack

- Creating a Tuple

- Operations on Tuples
- Tuple Functions

## Creating Tuples

```python
# empty
t1 = ()
print(t1)

# create a tuple with a single item
t2 = ('hello',)
print(t2)

# homo
t3 = (1,2,3,4)
print(t3)

# hetro
t4 = (1,2,5, True,[1,2,3])
print(t4)

# tuple
t5 = (1,2,3, (4,5))
print(t5)

# using type conversion
t6 = tuple('hello')
print(t6)
```

# Difference between Lists and Tuples

- Syntax
- Mutability
- Speed
- Memory
- Built in functionality
- Error prone
- Usability

```python
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)


start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
```

```python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

# Sets

A set is an unordered collection of items.every set element is unique (no duplicates) and must be immutable
(can not be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union,
intersection, symmetric difference, etc.

Characterstics:

- unordered

- Mutable

- No Duplicates

- Can't contain data types

## Creating Sets

```python
# empty
s = set()
print(s)
print(type(s))

# 1D and 2D
s1 = {1,2,3}
print(s1)

# s2 ={1,2,3,{4,5}}
# print(s2)
# homo and hetro
s3 = {1,'hello',4,5,(1,2,3)}
print(s3)

# using type conversion

s4 = set([1,2,3])
print(s4)

# duplicates not allowed
s5 = {1,2,3,4,1,2,3}
print(s5)

# set can't have mutable items
s6 = {1,2,[3,4]}
print(s6)
```

```python
s1 = {1,2,3}
s2 = {3,2,1}

print(s1 == s2)
```

# Set Operation

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1 | s2

# Union(|)
# intersection(&)

s1 & s2
# Difference(-)
s1 - s2
s2 - s1
# symmetric Difference(^)
s1 ^ s2

# Membership Test
1 not in s1

# Iteration
for i in s1:
  print(i)
```

#Set Functions

```python
# len/sum/min/max/sorted
S = {1,2,3,4,5}
print(len(S))
print(sum(S))
print(min(S))
print(max(S))
print(sorted(S))
print(sorted(S,reverse=True))
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2
s1.union(s1)
s1.update(s2)
print(s1)
print(s2)
```

```python
# intersection / intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference /difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)
s1.symmetric_difference_update(s2)
print(s1)
```

```python
# isdisjoint/issubset/ issuperset
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.isdisjoint(s2)
```

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.issubset(s2)
```

```python

s1 = {1,2,3,4,5}
s2 = {3,4,5}
s1.issuperset(s2)
```

```python
# copy
s1 = {1,2,3,4,5}
s2 = s1.copy()
print(s1)
print(s2)
```

# Dictionary

## Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

- In some languages it is known as map or assosiative arrays.


dict = {'name':'saif','age',33,'gendor':'male'}

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

# Create Dictionary

```python
# empty dictionary
d = {}
d

# 1D dictionary
d1 = {'name':'saif', 'gender':'male'}
print(d1)

# with mixed keys
d2 = {(1,2,3):1, 'hello':'world'}
print(d2)

# 2D dictionary -> JSON
s = {
    'name':'saif',
    'college':'bit',
    'sem':4,
    'subject':{
        'dsa': 50,
        'maths':67,
        'english':34
    }

}
print(s)

# using sequence and dict function
d4 = dict([('name','saif'),('age',32),(3,3)])
print(4)

# duplicate keys
d5 = {'name':'saif','name':'rahul'}
print(d5)
# mutable items as keys
d6 = {'name':'saif', (1,2,3):2}
print(d6)
```

## Dictionary Comprehension
image.png

```python
# print 1st 10 numbers and their squares
{i:i**2 for i in range(1,11)}
```

```python
distances = {'delhi':1000,'mumbai':2000,'bangalore':3000}
print(distances.items())
```

```python
# uisng existing dict
distances = {'delhi':1000, 'mumbai':2000, 'bangalore':3000}
{key:value*0.62 for (key,value) in distances.items()}
```

```python
# using zip
days = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
temp_c = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for (i,j) in zip(days,temp_c)}
```

```python
# using if condition
products = {'phone':10, 'laptop':0, 'charger':30, 'tablet':0}
{key:value for (key,value) in products.items() if value>0}
```

```python
# Nested Comprehension
# print table of number from 2 to 4
{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
```
