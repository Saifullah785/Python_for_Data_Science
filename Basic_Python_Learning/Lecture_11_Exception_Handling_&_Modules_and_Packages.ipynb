# Exeptions

if things go wrong during the execution of the program(runtime).it generally happens when something unforeseen has happpend.

- Exception are raised by Python runtime
- you have to takle is on the fly

# Examples
- Memory overflow
- Divide by o -> logical error

```python
# why is it important to handle exceptions
# how to handle exceptions
# -> try except block
```

```python
# let's create a file
with open('sample.txt','w') as f:
  f.write('hello world')
```

```python
# Try catch demo
try:
  with open('sample1.txt','r') as f:
    print(f.read())
except:
  print('sorry file not found')
```

```python
# catching specific exception
try:
  m=5
  f= open('sample1.txt','r')
  print(f.read())
  print(m)
  print(5/2)
  L = [1,2,3]
  L[100]
except FileNotFoundError:
  print('sorry file not found')
except NameError:
  print('variable not defined')
except ZeroDivisionError:
  print("can't divide by 0")
except Exception as e:
  print(e)
```

```python
# else
try:
  f = open('sample.txt', 'r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())
```
