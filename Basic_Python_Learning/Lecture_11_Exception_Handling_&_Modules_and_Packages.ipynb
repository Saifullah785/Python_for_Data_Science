# There are 2 stages where error may happen in a prgram

- During compilation -> Syntax Error

- During execution -> Exceptions

<div style="background-color:#EFF6FF; border:1px solid #BFDBFE; border-left:6px solid #2563EB; border-radius:12px; padding:18px 20px; margin:16px 0;">
  <h3 style="color:#1D4ED8; margin-top:0;">📘 Stage 1: The Code Assembly Line (Compilation vs. Execution)</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> Python processes your code in two major phases. First, it compiles your source code into bytecode, checking if you followed Python's grammatical rules. Second, it executes that bytecode inside the runtime environment.</p>
  <p style="color:#1E293B;">⚡ <strong>Key Distinction:</strong>
    <ul style="color:#1E293B;">
      <li><strong>Syntax Errors:</strong> Blockages at compile-time. Your code is unreadable to Python.</li>
      <li><strong>Exceptions:</strong> Speedbumps at runtime. Your syntax is perfect, but the instructions cannot be executed physically (e.g., searching for a file that doesn't exist).</li>
    </ul>
  </p>
</div>

  <h3 style="color:#065F46; margin-top:0;">🛡️ The Exception Shield: Try-Except Blocks</h3>
  <p style="color:#064E3B;"><strong>Definition:</strong> Instead of letting a runtime error instantly crash your software, Python provides the <code>try-except</code> block. This structure allows your program to intercept an exception, handle the issue gracefully, and continue executing.</p>
  <div style="background-color:#FFFFFF; border:1px solid #D1FAE5; border-radius:8px; padding:12px 14px; margin:12px 0;">
    <p style="color:#064E3B; margin-top:0;">⚙️ <strong>Mechanism:</strong></p>
    <ol style="color:#064E3B; margin-bottom:0;">
      <li>Code inside the <strong><code>try:</code></strong> block is executed sequentially.</li>
      <li>If an exception occurs, execution immediately stops and jumps to the matching <strong><code>except:</code></strong> block.</li>
      <li>If no exception occurs, the <strong><code>except:</code></strong> block is completely ignored.</li>
    </ol>
  </div>
</div>

  <h3 style="color:#92400E; margin-top:0;">🛡️ Leveling Up: Specific Exception Handling</h3>
  <p style="color:#78350F;">Using a bare, generic <code>except:</code> clause is highly discouraged in production because it catches every error blindly. Instead, explicitly declare the exact exceptions you expect to handle.</p>
  <p style="color:#78350F; margin-bottom:0;">💡 <strong>Pro Tip:</strong> Order exceptions from most specific (e.g., <code>FileNotFoundError</code>) to most generic (e.g., <code>Exception as e</code>) so the most accurate handler catches the target exception first.</p>
</div>

```python
# This code creates a file named 'sample.txt' and writes 'hello world' into it.
with open('sample.txt','w') as f:
  f.write('hello world')
```

```python
# This code demonstrates a basic try-except block to handle a FileNotFoundError.
try:
  with open('sample1.txt','r') as f:
    print(f.read())
except:
  print('sorry file not found')
```

```python
# This code demonstrates catching specific exceptions in a try-except block.
try:
  m=5
  f= open('sample1.txt','r')
  print(f.read())
  print(m)
  print(5/2)
  L = [1,2,3]
  L[100]
except FileNotFoundError:
  print('sorry file not found')
except NameError:
  print('variable not defined')
except ZeroDivisionError:
  print("can't divide by 0")
except Exception as e:
  print(e)
```

<div style="background-color:#F0FDF4; border:1px solid #BBF7D0; border-left:6px solid #16A34A; border-radius:12px; padding:18px 20px; margin:16px 0;">
  <h3 style="color:#166534; margin-top:0;">🍃 The Complete Exception Suite: Else & Finally</h3>
  <p style="color:#14532D;">Python provides two additional statements to construct highly robust error checking workflows:</p>
  <ul style="color:#14532D;">
    <li><strong><code>else:</code></strong> Code blocks that execute **only** if the <code>try</code> block ran successfully without raising any exceptions.</li>
    <li><strong><code>finally:</code></strong> Code blocks that **always** execute, regardless of whether an exception was raised, caught, or ignored. Perfect for cleanups (like closing open files or database connections).</li>
  </ul>
</div>

```python
# This code demonstrates the use of the 'else' block with try-except, which executes if no exception occurs.
try:
  f = open('sample.txt', 'r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())
```

```python
# This code demonstrates the 'finally' block, which always executes regardless of whether an exception occurred.
# else
try:
  f = open('sample1.txt','r')
except FileNotFoundError:
  print('file nai mili')
except Exception:
  print('kuch to lafda hai')
else:
  print(f.read())
finally:
  print('ye to print hoga hi')
```

```python
# This code defines a Bank class with a withdraw method that raises custom exceptions for invalid amounts or insufficient balance.
class bank:
  def __init__(self, balance):
    self.balance = balance
  def withdraw(self, amount):
    if amount < 0:
      raise Exception('amount cannot be -ve')
    if self.balance < amount:
      raise Exception('paise nai hai tere pass')
    self.balance = self.balance - amount

obj = bank(10000)
try:
  obj.withdraw(15000)
except Exception as e:
  print(e)
else:
  print(obj.balance)
```

<div style="background-color:#EFF6FF; border:1px solid #BFDBFE; border-left:6px solid #2563EB; border-radius:12px; padding:18px 20px; margin:16px 0;">
  <h3 style="color:#1D4ED8; margin-top:0;">🏗️ Building Custom Exception Classes</h3>
  <p style="color:#1E293B;"><strong>Definition:</strong> By subclassing Python's base <code>Exception</code> class, you can create tailored exceptions that carry distinct domain definitions for your systems (e.g., <code>SecurityError</code>, <code>InsufficientBalanceError</code>).</p>
  <p style="color:#1E293B; margin-bottom:0;">💡 <strong>Takeaway:</strong> Inheriting from <code>Exception</code> guarantees that your custom exception class integrates natively with pythonic <code>try-except</code> architecture blocks.</p>
</div>

```python
# This code defines a custom exception class MyException and demonstrates its use in a Bank class's withdraw method.
class MyException(Exception):
  def __init__(self, message):
    print(message)

class Bank:
  def __init__(self, balance):
    self.balance = balance

  def withdraw(self, amount):
    if amount < 0:
      raise MyException('amount cannot be -ve')
    if self.balance < amount:
      raise MyException('paise nai hai tere pass')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(5000)
except MyException as e:
  pass
else:
  print(obj.balance)
```

```python
# This code defines a custom SecurityError exception and a Google class to demonstrate login with security checks.
class SecurityError(Exception):
  def __init__(self, message):
    print(message)
  def logout(self):
    print('logout')

class Google:
  def __init__(self, name, email, password, device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self, email, password, device):
    if device != self.device:
      raise SecurityError('bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')

obj = Google('saif','saif@gmail.com','1234','android')

try:
  obj.login('saif@gmail.com','1234','android')
except SecurityError as e:
  e.logout()

else:
  print(obj.name)
finally:
  print('database connection closed')
```
