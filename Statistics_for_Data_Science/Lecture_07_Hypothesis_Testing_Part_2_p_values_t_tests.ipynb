```python
from scipy.stats import t

# Set the t-value and degrees of freedom
t_value = -5.25

# Replace this with your specific degrees of freedom
df = 58

# Calculate the CDF value
cdf_value = t.cdf(t_value, df)
print(cdf_value)

```

```python
from scipy.stats import shapiro

destop_users = [12, 15, 18, 16, 20, 17, 14, 22, 19, 21, 23, 18, 25, 17, 16, 24, 20, 19, 22, 18, 15, 14, 23, 16, 12, 21, 19, 17, 20, 14]

mobile_users = [10, 12, 14, 13, 16, 15, 11, 17, 14, 16, 18, 14, 20, 15, 14, 19, 16, 15, 17, 14, 12, 11, 18, 15, 10, 16, 15, 13, 16, 11]

# Perform the shapiro-wilk test for both destop and mobile users
shapiro_desktop = shapiro(destop_users)
shapiro_mobile = shapiro(mobile_users)

print("Shapiro-wilk test for destop users:", shapiro_desktop)
print("Shapiro-wilk test for mobile users:", shapiro_mobile)
```

```python
from scipy.stats import levene

destop_users = [12, 15, 18, 16, 20, 17, 14, 22, 19, 21, 23, 18, 25, 17, 16, 24, 20, 19, 22, 18, 15, 14, 23, 16, 12, 21, 19, 17, 20, 14]

mobile_users = [10, 12, 14, 13, 16, 15, 11, 17, 14, 16, 18, 14, 20, 15, 14, 19, 16, 15, 17, 14, 12, 11, 18, 15, 10, 16, 15, 13, 16, 11]

# Perform Levene's test
levene_test = levene(destop_users, mobile_users)

print("Levene's test:", levene_test)
```
